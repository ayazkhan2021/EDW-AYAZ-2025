```python
def lesser_of_two_evens(a,b):
    result = 0
    if a%2 == 0 and b%2 == 0:
      if a>b:
        result = b
      else:
        result = a
    elif a%2 != 0 or b%2 != 0:
      if a>b:
        result = a
      else:
        result = b
    return result
```

```python
lesser_of_two_evens(2,5)
```

```python
def animal_crackers(input_str):
  my_list = input_str.split()
  if my_list[0][0] == my_list[1][0]:
    return True
  else:
    return False
```

```python
def makes_twenty(n1,n2):
    sum_of_numbers = n1 + n2
    if sum_of_numbers == 20 or n1 == 20 or n2 == 20:
      return True
    else:
      return False
```

```python
makes_twenty(1, 5)
```

```python
def old_macdonald(name):
    result = []
    position_of_letter = 1
    for letter in name:
      if position_of_letter == 1 or position_of_letter == 4:
        result.append(letter.upper())
      else:
        result.append(letter)

      position_of_letter += 1

    result = "".join(result)

    return result
```

```python
old_macdonald('macdonald')
```

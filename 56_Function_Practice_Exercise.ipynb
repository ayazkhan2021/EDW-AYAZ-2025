```python
def lesser_of_two_evens(a,b):
    result = 0
    if a%2 == 0 and b%2 == 0:
      if a>b:
        result = b
      else:
        result = a
    elif a%2 != 0 or b%2 != 0:
      if a>b:
        result = a
      else:
        result = b
    return result
```

```python
lesser_of_two_evens(2,5)
```

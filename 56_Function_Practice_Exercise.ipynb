```python
def lesser_of_two_evens(a,b):
    result = 0
    if a%2 == 0 and b%2 == 0:
      if a>b:
        result = b
      else:
        result = a
    elif a%2 != 0 or b%2 != 0:
      if a>b:
        result = a
      else:
        result = b
    return result
```

```python
lesser_of_two_evens(2,5)
```

```python
def old_macdonald(name):
    result = []
    position_of_letter = 1
    for letter in name:
      if position_of_letter == 1 or position_of_letter == 4:
        result.append(letter.upper())
      else:
        result.append(letter)

      position_of_letter += 1

    result = "".join(result)

    return result
```

```python
old_macdonald('macdonald')
```

```python
def master_yoda(input_str):
  myList = input_str.split()
  result = []
  for word in reversed(myList):
    result.append(word)

  return " ".join(result)
```

```python
master_yoda('We are ready')
```

```python
def has_33(nums):
  first_3 = nums.index(3)

  search_till = first_3 +1
  search_start = first_3 + 1

  while search_start <= search_till:
    if nums[search_start] == 3:
      return True
    else:
      return False
    search_start += search_start



```

```python
has_33([1, 3, 3])
```

```python
def paper_doll(text):
    myList = []
    for letter in text:
      for i in range(3):
        myList.append(letter)

    return "".join(myList)
```

```python
def blackjack(a,b,c):
    result = a + b + c
    if result <= 21:
      return result
    elif result > 21 and (a==11 or b==11 or c==11):
      result = result - 10
      return result
    elif result > 21:
      return 'BUST'
```

```python
import math

def volume_of_sphere(radius):
  """Calculates the volume of a sphere given its radius."""
  return (4/3) * math.pi * (radius**3)
```

```python
# Example usage:
radius = 2
volume = volume_of_sphere(radius)
print(f"The volume of a sphere with radius {radius} is {volume}")
```

```python
def up_low(s):
  """Counts the number of uppercase and lowercase letters in a string."""
  lower = 0
  upper = 0
  for letter in s:
    if letter.islower():
      lower = lower + 1
    elif letter.isupper():
      upper = upper + 1

  print(f'Number of uppercase leter is: {upper}' )
  print(f'Number of lowercase leter is: {lower}' )
```

```python
up_low("Hello Mr. Rogers, how are you this fine Tuesday?")
```

```python
def ispangram(str1, alphabet=string.ascii_lowercase):
  input_str = set(str1.lower().replace(" ", ""))
  string2 = list(alphabet)
  input_str = sorted(input_str)
  if string2 == input_str:
    return True
  else:
    return False
```

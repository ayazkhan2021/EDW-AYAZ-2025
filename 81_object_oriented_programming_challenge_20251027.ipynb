```python
class Account():

  def __init__(self, owner, balance = 0):
    self.owner = owner
    self.balance = balance

  def deposit(self, deposit):
    self.balance = self.balance + deposit
    print(f"After deposit your new balance is ${self.balance}")

  def withdrawal(self, amount):
    if amount > self.balance:
      print(f'You cannot withdraw more than your balance')
    else:
      self.balance = self.balance - amount
      print(f"After withdrawal your new balance is ${self.balance}")

  def __str__(self):
    return f"Account owner: {self.owner} \nAccount Balance: ${self.balance}"
```

```python
acc1 = Account('Jose', 229)
```

```python
print(acc1)
```

```python
acc1.deposit(600)
```

```python
print(acc1)
```

```python
acc1.withdrawal(200)
```

```python
acc1.withdrawal(1400)
```

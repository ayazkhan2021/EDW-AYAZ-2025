```python
def replay():
  choice = input('Play again? "Yes" or "No": ')
  return choice == 'Yes'
```

```python
def player_input():
  marker = ''
  while not (marker == 'X' or marker == 'O'):
    marker = input('Player 1: Choose X or O: ').upper()
    if marker == 'X':
      return ('X', 'O')
    else:
      return ('O', 'X')
```

```python
import random

def choose_first():
  flip = random.randint(0, 1)
  if flip == 0:
    return 'Player 1'
  else:
    return 'Player 2'
```

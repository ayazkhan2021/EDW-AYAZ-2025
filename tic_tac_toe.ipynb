```python
def replay():
  choice = input('Play again? "Yes" or "No": ')
  return choice == 'Yes'
```

```python
def player_input():
  marker = ''
  while not (marker == 'X' or marker == 'O'):
    marker = input('Player 1: Choose X or O: ').upper()
    if marker == 'X':
      return ('X', 'O')
    else:
      return ('O', 'X')
```
